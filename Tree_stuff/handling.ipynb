```python
from Bio import Phylo

tree = Phylo.read('/home/school/masters/Scripts/Tree_stuff/treeoflife.4.0.family.tree', 'newick')
tree.ladderize()
Phylo.draw(tree)

families = set()
orders = set()
to_sort = set()
for clade in tree.find_clades():
    name = clade.name
    if name:
        if name.endswith('ales'):
            orders.add(name)
        elif name.endswith('aceae'):
            families.add(name)
        else:
            to_sort.add(name)

print(len(families))
print(len(orders))
print(len(to_sort))

print(families)

families = [family.split('_')[1] for family in families]
print(families)
```
