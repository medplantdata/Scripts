making a test set that is half of coconut and a ref database that is just one entry for the normalizer testing

```python
import pandas as pd

df = pd.read_csv('/home/school/masters/Scripts/coconut_full/full_coconut_for_DB.csv')

print(df.shape[0])
print(df.columns)

if df['canonical_smiles'].isna().any():
    print('fuck')
# fixing isomeric/canon smiles thing

df['isomeric_smiles'] = df['isomeric_smiles'].fillna(df['canonical_smiles'])
df.drop(columns = ['canonical_smiles'], inplace = True)
df = df.rename(columns = {'isomeric_smiles': 'smiles'})

print(df.columns)

new = df.loc[df['compound_id'].between(1,40000)].copy()
wand = df.loc[df['compound_id'] == 1].copy()

new = new.rename(columns={'compound_id':'id','iupac':'n_iupac','synonyms':'n_synonyms','name':'n_name', 'collections': 'n_collections', 'dois': 'n_dois','cid': 'n_cid', 'cas': 'n_cas'})



new.to_csv('/home/school/masters/Scripts/testing_compound_normalizer/coconut_test.csv', index  = False)
wand.to_csv('/home/school/masters/Scripts/testing_compound_normalizer/wand_test.csv', index = False)
```

gonna send all the coconut compounds associated with a plant to wonko to the normalizaaaa

```python
import pandas as pd

plant_compounds = pd.read_csv('/home/school/masters/Scripts/coconut_full/coconut_plants/checklist_matched_organisms.tsv', sep = '\t')

print(plant_compounds.columns)

plant_compounds = plant_compounds[['original_coconut_id','kingdom']]

plant_only =  plant_compounds.loc[plant_compounds['kingdom'] == 'Plantae'].copy()

plant_set = set(plant_only['original_coconut_id'].to_list())

df = pd.read_csv('/home/school/masters/Scripts/coconut_full/coconut_with_cids.csv',usecols=['identifier','canonical_smiles','standard_inchi','standard_inchi_key','name','iupac_name','np_likeness','collections','dois','synonyms','cas','coconut_id','cid'])

df = df.rename(columns={'identifier':'id', 'canonical_smiles': 'smiles', 'standard_inchi': 'inchi', 'standard_inchi_key': 'inchi_key', 'name' : 'n_name', 'iupac_name': 'n_iupac', 'np_likeness': 'n_np_likeness', 'collections': 'n_collections', 'dois': 'n_dois', 'synonyms': 'n_synonyms', 'cas': 'n_cas','coconut_id': 'n_coconut_id', 'cid': 'n_cids'})

df = df.loc[df['id'].isin(plant_set)].copy()

df.to_csv('/home/school/masters/Scripts/coconut_full/coconut_for_normalizer.csv',index=False)

```
