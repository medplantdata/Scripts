# Trying to make a nice interconverter for formats and such

### inchi to iupac with nispo

```python
from rdkit import Chem
from rdkit.Chem import inchi
from nispo import mol_to_iupac

mol = inchi.MolFromInchi('InChI=1S/C30H37N3O5/c1-4-36-25-15-10-8-13-21(25)29-28-22(20-12-7-9-14-23(20)31-28)18-24-30(35)32(19-26(34)33(24)29)17-11-16-27(37-5-2)38-6-3/h7-10,12-15,24,27,29,31H,4-6,11,16-19H2,1-3H3/t24-,29?/m0/s1')
mol_to_iupac(mol)
```
